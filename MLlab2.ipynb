```python
df = pd.DataFrame({
    'Age': [25, 30, np.nan, 40, 1000, 25],
    'Salary': [50000, 60000, 55000, np.nan, 70000, 50000],
    'Gender': ['M', 'F', 'F', 'M', np.nan, 'M'],
    'Purchased': [0, 1, 0, 1, 0, 0]
})
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].max(), inplace=True)
```

```python
df
```

```python
df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)
```

```python
df
```

```python
df['Salary_log']= np.log(df['Salary' ])
df
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(-1, 1))
df[['Age', 'Salary', 'Salary_log']] = scaler.fit_transform(df[['Age', 'Salary','Salary_log']])

df
```

```python
df['Purchased'].value_counts()
```

```python
from imblearn.under_sampling import RandomUnderSampler

X = df.drop('Purchased', axis=1)
y = df['Purchased']

rus = RandomUnderSampler()
X_under, y_under = rus.fit_resample(X, y)
X_under
```

```python
from sklearn.preprocessing import LabelEncoder
from imblearn.over_sampling import SMOTE
x=df.drop('Purchased',axis=1)
y=df['Purchased']
le=LabelEncoder()
x['Gender']=le.fit_transform(x['Gender'])
smote=SMOTE(k_neighbors=1 , random_state=42)
x_smote,y_smote=smote.fit_resample(x,y)
```
